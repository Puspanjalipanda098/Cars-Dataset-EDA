```python
df.drop_duplicates(inplace=True)
```

```python
#Outliers
plt.figure(figsize=(10,5))
sns.boxplot(df['brand'])
plt.show
```

```python
df.select_dtypes(include = 'number').plot(kind='box',figsize=(12,6),rot=45)
plt.show()
```

```python
#Universal Analysis
sns.histplot(df['brand'],kde=True)
plt.title('brand distribution')
plt.show()
```

```python
sns.boxplot(x='selling_price',hue='brand',data=df)
plt.title("BRAND VS SELLING_PRICE")
plt.show()
```

```python
#Bivariate Analysis
sns.regplot(x='selling_price',y='km_driven',data=df)
plt.title("SELLING_PRICE VS KM_DRIVEN")
plt.show()
```

```python
#Multivariate Analysis
plt.figure(figsize=(12,6))
sns.scatterplot(x='selling_price',y='km_driven',hue='brand',data=df)
plt.title("SELLING_PRICE VS KM_DRIVEN")
plt.show()
```
